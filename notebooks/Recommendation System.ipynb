```python
import pandas as pd
```

```python
df = pd.read_csv('segmented_customers.csv')
print(df.head())
```

```python
recommendations = {
    "High Value Customers": {
        "Strategy": "Loyalty Rewards & Premium Offers",
        "Channels": ["Email", "Personal Calls"],
        "Offers": ["Exclusive Discounts", "Early Access"],
        "Goal": "Increase retention and lifetime value"},
    
    "New Customers": {
        "Strategy": "Welcome Campaign",
        "Channels": ["Email", "SMS"],
        "Offers": ["First Purchase Discount"],
        "Goal": "Convert into repeat customers"},
    
    "At Risk Customers": {
        "Strategy": "Re-engagement Campaign",
        "Channels": ["Email", "Push Notifications"],
        "Offers": ["Win-back Discounts"],
        "Goal": "Prevent churn"},
    
    "Budget Buyers": {
        "Strategy": "Price-Based Promotions",
        "Channels": ["SMS", "Ads"],
        "Offers": ["Coupons", "Seasonal Discounts"],
        "Goal": "Increase purchase frequency"}
}
```

```python
df.rename(columns={'Clusters': 'Cluster'}, inplace=True)
```

```python
def get_recommendation(segment):
    return recommendations.get(segment, "No recommendation available")
```

```python
df['Recommendation'] = df['Segment'].apply(get_recommendation)
```

```python
df.head()
```

```python
df['Strategy'] = df['Segment'].apply(
    lambda x: recommendations[x]['Strategy'] if x in recommendations else "Unknown")

df['Channels'] = df['Segment'].apply(
    lambda x: ", ".join(recommendations[x]['Channels']) if x in recommendations else "Unknown")

df['Offers'] = df['Segment'].apply(
    lambda x: ", ".join(recommendations[x]['Offers']) if x in recommendations else "Unknown")

df['Goal'] = df['Segment'].apply(
    lambda x: recommendations[x]['Goal'] if x in recommendations else "Unknown")

df.head()
```

```python
df.to_csv('Recommendation System.csv', index=False)
```
